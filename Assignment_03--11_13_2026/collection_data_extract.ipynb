### collection_data_extract.py

This file runs the operations outlined in part 1 of assignment 3: extracting entire collection data from the selected collection. It extracts data from the library of congress website using the HTTP requests Python module.
Originally written as a regular python script, I have chosen to rewrite it as a jupyter notebook.

The collection the extraction is being run on can be faound at this link: https://www.loc.gov/free-to-use/colors-tell-the-story/

#### Imports

```python
import csv
import json

import requests
```

#### Extract Collection Data

```python
# setting up request:

baseURL = 'https://www.loc.gov'
endpoint = '/free-to-use'
collection = '/colors-tell-the-story'
parameters = {
    'fo' : 'json'
}

# running request & saving as "collection_list_response":

collection_list_response = requests.get(baseURL + endpoint + collection, params=parameters)
```

```python
# checking for success / errors:

print(collection_list_response.url)
print(collection_list_response.status_code)
```

```python
# saving data as a json object:

collection_json = collection_list_response.json()
```

```python
# testing success of metadata extraction

print(json.dumps(collection_json, indent=4))
print(collection_json.keys())
```

#### Getting Individual Collection Items from JSON

```python
# saving collection item list section of json to a separate variable

collection_items = collection_json['content']['components'][0]['masonry_gallery'][0]['items']
```

```python
# testing that items are correctly returned, type of object new variable is, and how many items are saved to the list

print(collection_items)
print(type(collection_items))
print(len(collection_items))
```

#### Saving Colleciton Item List to CSV File

```python
filename = 'collection_item_list.csv'
headers = ['alt', 'image', 'link', 'title']

with open(filename, 'w', encoding='utf-8', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=headers)
    writer.writeheader()
    for item in collection_items:
        item['title'] = item['title'].strip()
        writer.writerow(item)
    print('wrote', filename)
```
